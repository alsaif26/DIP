# 05 Negative Transformation
প্রতিটি Run করলে আলাদাভাবে image upload করা যাবে।

```python
from google.colab import files
import cv2
import matplotlib.pyplot as plt

file = files.upload()
name = list(file.keys())[0]

image = cv2.imread(name, 0)
negative = 255 - image

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(negative, cmap="gray")
plt.title("Negative")
plt.axis("off")

plt.show()
```
