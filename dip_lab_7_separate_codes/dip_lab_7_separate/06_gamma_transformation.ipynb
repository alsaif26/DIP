# 06 Gamma Transformation
প্রতিটি Run করলে আলাদাভাবে image upload করা যাবে।

```python
from google.colab import files
import cv2
import matplotlib.pyplot as plt

file = files.upload()
name = list(file.keys())[0]

image = cv2.imread(name, 0)

gamma1 = 0.5
gamma2 = 2

image1 = 255 * (image / 255) ** gamma1
image2 = 255 * (image / 255) ** gamma2

plt.subplot(1, 3, 1)
plt.imshow(image, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(image1, cmap="gray")
plt.title("Gamma 0.5")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(image2, cmap="gray")
plt.title("Gamma 2")
plt.axis("off")

plt.show()
```
