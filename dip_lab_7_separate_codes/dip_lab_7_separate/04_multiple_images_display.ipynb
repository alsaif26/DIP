# 04 Multiple Images Display
প্রতিটি Run করলে আলাদাভাবে image upload করা যাবে।

```python
from google.colab import files
import cv2
import matplotlib.pyplot as plt

file = files.upload()
name = list(file.keys())[0]

image = cv2.imread(name)
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
negative = 255 - gray

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(gray, cmap="gray")
plt.title("Gray")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(negative, cmap="gray")
plt.title("Negative")
plt.axis("off")

plt.show()
```
