# 07 Brightness Comparison
প্রতিটি Run করলে আলাদাভাবে image upload করা যাবে।

```python
from google.colab import files
import cv2
import matplotlib.pyplot as plt
import numpy as np

file = files.upload()
name = list(file.keys())[0]

image = cv2.imread(name, 0)
negative = 255 - image
gamma1 = 255 * (image / 255) ** 0.5
gamma2 = 255 * (image / 255) ** 2

a = np.mean(image)
b = np.mean(negative)
c = np.mean(gamma1)
d = np.mean(gamma2)

print("Original:", a)
print("Negative:", b)
print("Gamma 0.5:", c)
print("Gamma 2:", d)

plt.bar(["Original", "Negative", "Gamma 0.5", "Gamma 2"], [a, b, c, d])
plt.ylabel("Brightness")
plt.title("Brightness Comparison")
plt.show()
```
